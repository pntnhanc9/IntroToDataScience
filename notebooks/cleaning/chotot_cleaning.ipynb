```python
import pandas as pd
import csv
```

```python
df = pd.read_csv('../../data/raw/chotot_raw.csv')
print(df['Size'].dtype)
new_df = pd.DataFrame({
    'url': df['IRL (Link gốc)'],
    'title': df['Name'],
    'price': df['Money (Price)'].astype(float),
    'area': df['Size'].str.replace(" m²", "").astype(float),
    'address': df['Address'],
    'description': (
        df['Body'].fillna('') +
        ", tiền cọc: " + df['Deposit'].fillna('') +
        ", tình trạng nội thất: " + df['Furniture'].fillna('')
    ),
    'source': df['Web']
})
```

```python
# Xuất CSV chuẩn để không bị mất nội dung
new_df.to_csv(
    '../../data/clear/chotot_clear.csv',
    index=False,
    quoting=csv.QUOTE_ALL,      # quan trọng!
    escapechar='\\',            # escape ký tự "
    encoding='utf-8-sig'
)
```
