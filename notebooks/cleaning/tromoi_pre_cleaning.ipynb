```python
# Convert area text to numeric (m²)
import pandas as pd

# Load dataset (keep only for context; rest removed)
df = pd.read_csv('data_tromoi.csv')

# Normalize and extract numeric value from strings like "25m²", "30 m2", "20,5 m²"
df['area'] = (
    df['area'].astype(str)
      .str.replace(',', '.', regex=False)              # handle decimal comma
      .str.extract(r'(\d+(?:\.\d+)?)')[0]           # capture first numeric token
      .astype(float)
)

```

```python
# Save cleaned dataset to a new CSV file
output_path = 'tromoi_clear.csv'
df.to_csv(output_path, index=False, encoding='utf-8-sig')
print(f"Saved cleaned dataset to {output_path} with shape {df.shape}")
```
