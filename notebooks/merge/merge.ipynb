```python
import pandas as pd
import csv

df1 = pd.read_csv('../../data/processed/chotot_clear.csv')
df2 = pd.read_csv('../../data/processed/tromoi_clear.csv')
df3 = pd.read_csv('../../data/processed/homedy_clear.csv')
df3.drop('index', axis=1, inplace=True)
df4 = pd.read_csv('../../data/processed/phongtro123_clear.csv')

```

```python
df_combined = pd.concat([df1, df2,df3,df4], ignore_index=True)
df_combined.to_csv(
    'data.csv',
    index=False,
    quoting=csv.QUOTE_ALL,      # quan trọng!
    escapechar='\\',            # escape ký tự "
    encoding='utf-8-sig'
)
```
